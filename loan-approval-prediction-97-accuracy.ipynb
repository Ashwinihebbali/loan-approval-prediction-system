## 5. Outlier Detection using IQR Method
Identify outliers in numeric columns using the Interquartile Range (IQR) method to understand data anomalies.

```python
for col in numeric_col:
    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)
    IQR = Q3 - Q1
    upper_bound = Q3 + 1.5 * IQR
    Lower_bound = Q1 - 1.5 * IQR

    outliers = data[(data[col]< Lower_bound) | (data[col]>upper_bound)]
    print(f"column: {col}")
    print(f"Outliers : {outliers.shape[0]}")
```

## 7. Apply Box-Cox Transformation
Apply Box-Cox transformation to normalize the distribution of asset values and handle outliers.

```python
for col in outliers_col:
    Q1 = data[f"{col}_boxcox"].quantile(0.25)
    Q3 = data[f"{col}_boxcox"].quantile(0.75)
    IQR = Q3 - Q1
    upper_bound = Q3 + 1.5 * IQR
    Lower_bound = Q1 - 1.5 * IQR

    outliers = data[(data[f"{col}_boxcox"]< Lower_bound) | (data[f"{col}_boxcox"]>upper_bound)]
    print(f"column: {col}")
    print(f"Outliers : {outliers.shape[0]}")
    
```

```python
x = data_encode.drop(columns=['loan_id', 'self_employed_ No', 'self_employed_ Yes','education_ Not Graduate','education_ Graduate','loan_status_ Approved','loan_status_ Rejected'], axis=1)
x.head()
```

## 13. Prepare Features and Target for Model Training
Prepare feature matrix (X) by removing redundant columns and encode the target variable (y) for model training.

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder


le = LabelEncoder()

y = pd.Series(le.fit_transform(data['loan_status']))


print(y.head())
```

## 14. Split Data into Training and Testing Sets
Divide the data into training (80%) and testing (20%) sets for model development and evaluation.

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

## Scaling 

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

## 15. Train Logistic Regression Model
Build and train a Logistic Regression classifier to predict loan approval.

```python
Logistic_model = LogisticRegression(max_iter=1000)
model = Logistic_model.fit(X_train,y_train)
```

## 16. Make Predictions with Logistic Regression
Generate predictions on the test set using the trained Logistic Regression model.

```python
y_pred = model.predict(X_test)
```

## 17. Evaluate Logistic Regression on Test Set
Generate classification report to evaluate model performance on the test data.

```python
report = classification_report(y_test,y_pred)
print(f"{report}")
```

## 18. Evaluate Logistic Regression on Training Set
Generate classification report to evaluate model performance on the training data.

```python
y_pred_train = model.predict(X_train)
report = classification_report(y_train,y_pred_train)
print(f"{report}")
```

## 19. Train Decision Tree Model
Build and train a Decision Tree classifier with a maximum depth of 5 for interpretability.

```python
decision_tree_model = DecisionTreeClassifier(random_state=42,max_depth=5)
decision_tree_model.fit(X_train, y_train)
Y_Pred_DT = decision_tree_model.predict(X_test)
```

## 20. Evaluate Decision Tree on Test Set
Generate classification report to evaluate Decision Tree model performance on test data.

```python
report = classification_report(y_test,Y_Pred_DT)
print(f"{report}")
```

## 21. Evaluate Decision Tree on Training Set
Generate classification report to evaluate Decision Tree model performance on training data.

```python
Y_Pred_DT = decision_tree_model.predict(X_train)
report = classification_report(y_train,Y_Pred_DT)
print(f"Report on Training {report}")
```

## 22. Cross-Validation Analysis
Perform 5-fold cross-validation on the Decision Tree model to assess model stability and reliability across different data splits.

```python
cross_validation = cross_val_score(decision_tree_model,x,y,cv=5, scoring='accuracy')
print(f"Average Accuracy is: {cross_validation.mean()}")
```
